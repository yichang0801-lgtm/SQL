# SQL Toolkit

## SELECT ... FROM

**Purpose:** Choose which columns to view and from which table.

**Excel Equivalent:** Selecting columns

**Example**

```sql
SELECT customer_name, spend_hkd
FROM sales;
```

---

## WHERE

**Purpose:** Filter rows.

**Excel Equivalent:** Filter

**Example**

```sql
SELECT *
FROM sales
WHERE store_id = 1;
```

---

## ORDER BY ... LIMIT

**Purpose:** Sort results and return the top rows.

**Excel Equivalent:** Sort descending + Top N

**Example**

```sql
SELECT *
FROM sales
ORDER BY spend_hkd DESC
LIMIT 10;
```

---

## GROUP BY

**Purpose:** Aggregate data by category.

**Excel Equivalent:** Pivot Table

**Example**

```sql
SELECT store_id,
       SUM(spend_hkd) AS total_sales
FROM sales
GROUP BY store_id;
```

---

## JOIN ... ON

**Purpose:** Connect tables using a key.

**Excel Equivalent:** VLOOKUP / Merge

**Example**

```sql
SELECT *
FROM sales
JOIN customers
ON sales.customer_id = customers.customer_id;
```

---

# Quick Reference

| Clause | Function | Excel Equivalent |
|----------|----------|----------|
| SELECT ... FROM | choose columns from a table | select columns |
| WHERE | filter rows | filter |
| ORDER BY | sort data | sort |
| LIMIT | keep top N rows | Top N |
| GROUP BY | aggregate by group | pivot table |
| JOIN ... ON | connect tables using keys | VLOOKUP / Merge |

---

# SQL Notes - Common Errors

## 1. Table Does Not Exist

### Error

```text
Catalog Error: Table with name sale does not exist!
```

### Meaning

Table name is misspelled.

### Check

- Is the table name correct?
- Is it `sales` instead of `sale`?

---

## 2. Column Not Found

### Error

```text
Binder Error: Referenced column "spend" not found
```

### Meaning

The column name is misspelled, or the column exists in another table.

### Check

- Verify spelling
- Check table schema
- Confirm the correct table

---

## 3. Syntax Error

### Error

```text
Parser Error: syntax error at or near ...
```

### Meaning

Usually caused by a missing:

- comma (,)
- bracket ()
- quote (' ')

### Check

Look immediately before the highlighted word.

---

## 4. Double Quotes Instead of Single Quotes

### Error

```text
Binder Error: Referenced column "Electronics" not found
```

### Meaning

SQL interprets text inside double quotes as a column name.

### Wrong

```sql
WHERE category = "Electronics"
```

### Correct

```sql
WHERE category = 'Electronics'
```

### Rule

- Text values → single quotes `' '`
- Column names → no quotes (normally)

---

## 5. Missing GROUP BY

### Error

```text
column "date" must appear in the GROUP BY clause
```

### Meaning

When using aggregation, every selected column must be:

- aggregated
- or included in GROUP BY

### Wrong

```sql
SELECT date,
       SUM(spend_hkd)
FROM sales;
```

### Correct

```sql
SELECT date,
       SUM(spend_hkd)
FROM sales
GROUP BY date;
```

---

## 6. Aggregate Function Inside WHERE

### Error

```text
WHERE clause cannot contain aggregates
```

### Meaning

Aggregate results must be filtered with HAVING, not WHERE.

### Wrong

```sql
SELECT store_id,
       SUM(spend_hkd)
FROM sales
WHERE SUM(spend_hkd) > 1000
GROUP BY store_id;
```

### Correct

```sql
SELECT store_id,
       SUM(spend_hkd)
FROM sales
GROUP BY store_id
HAVING SUM(spend_hkd) > 1000;
```

### Rule

```text
WHERE  -> filter rows
HAVING -> filter groups
```

---

## 7. NULL Comparison

### Wrong

```sql
WHERE customer_id = NULL
```

### Meaning

NULL is not a value.

### Correct

```sql
WHERE customer_id IS NULL
```

### Or

```sql
WHERE customer_id IS NOT NULL
```

### Rule

Never use:

```sql
= NULL
```

Always use:

```sql
IS NULL
IS NOT NULL
```

---

# JOIN (Joining Two Tables)

A `JOIN` matches two tables using a shared column (key).

| SQL | pandas | R |
|------|------|------|
| `sales JOIN products ON sales.product_id = products.product_id` | `sales.merge(products, on="product_id")` | `inner_join(sales, products, by = "product_id")` |

## Common Keys

| Table | Key |
|---------|---------|
| sales ↔ products | `product_id` |
| sales ↔ stores | `store_id` |
| sales ↔ customers | `customer_id` |

## Example

```sql
SELECT *
FROM sales
JOIN products
ON sales.product_id = products.product_id;
```

## Business Logic

| Information Needed | Table |
|---------|---------|
| Product category | `products` |
| Store type | `stores` |
| Customer details | `customers` |
| Transaction data | `sales` |

## Mental Model

```text
sales
+
products
↓
JOIN ON product_id
↓
sales with category information
```

# Matching Two Lists (JOIN)

## How Matching Works

- `sales` knows the `product_id`
- `products` knows the product category
- `JOIN` matches the two tables using a shared key

```text
sales.product_id
=
products.product_id
```

---

## INNER JOIN

Keep only rows that match in both tables.

```text
sales ∩ products
```

- Unmatched rows are excluded
- Only records with matches on both sides remain

---

## LEFT JOIN

Keep every row from the first (left) table.

```text
sales ⟕ products
```

- Keep all rows from `sales`
- If no match exists in `products`, fill with `NULL`
- Missing matches do not remove rows from the left table

---

## Validation Rule

After a JOIN, always check:

```text
Does the total still add up?
```

If totals change unexpectedly:

- Rows may have been lost (`INNER JOIN`)
- Rows may have been duplicated
- Join key may be incorrect

# WITH (Common Table Expression / CTE)

## Purpose

Create a temporary named table, then query it.

```sql
WITH member_spend AS (
    SELECT customer_id,
           SUM(spend_hkd) AS total
    FROM sales
    GROUP BY customer_id
)
```

Then query:

```sql
SELECT *
FROM member_spend;
```

---

## Pattern

```sql
WITH table_name AS (
    query
)
SELECT ...
FROM table_name;
```

---

## Why Use It?

- Break a complex problem into smaller steps
- Improve readability
- Reuse intermediate results
- Read top-to-bottom

---

## Example

Step 1: Build a table of each customer's spending.

```sql
WITH member_spend AS (
    SELECT customer_id,
           SUM(spend_hkd) AS total
    FROM sales
    GROUP BY customer_id
)
```

Step 2: Ask questions about that table.

```sql
SELECT *
FROM member_spend
ORDER BY total DESC;
```

---

## Mental Model

```text
Raw Data
    ↓
WITH (temporary table)
    ↓
Analysis
    ↓
Answer
```

# Interview Question: "Your report's total doesn't match Finance's. What do you check?"

## Q1. Did a JOIN drop rows?

### Check

Compare row counts and totals before and after the JOIN.

### Why?

An `INNER JOIN` may remove records with no matching key.

### Example

```sql
SELECT COUNT(*)
FROM sales;
```

vs

```sql
SELECT COUNT(*)
FROM sales
INNER JOIN stores
ON sales.store_id = stores.store_id;
```

### Answer

A JOIN may have removed valid records.

---

## Q2. Are there NULL values that were skipped or miscounted?

### Check

Look for missing values.

### Example

```sql
WHERE store_type IS NULL
```

### Why?

Missing values may be excluded from summaries or grouped separately.

### Answer

NULL records may contain missing revenue or transactions.

---

## Q3. Is a WHERE filter removing more data than intended?

### Check

Review all filtering conditions.

### Example

```sql
WHERE spend_hkd > 5000
```

or

```sql
WHERE customer_id IS NOT NULL
```

### Why?

A filter may accidentally exclude valid records.

### Answer

The report and Finance may be using different filtering rules.

---

## Q4. Are both reports using the same date range and the same shops?

### Check

Compare:

- Date range
- Store list
- Customer scope

### Example

```text
Report A:
2026-10-01 to 2026-12-31

Report B:
2026-10-20 to 2026-12-31
```

### Why?

Totals are not comparable unless they use the same population.

### Answer

The numbers may differ because they are measuring different data sets.

---

## Validation Checklist

1. Compare row counts before and after JOIN.
2. Check for NULL values.
3. Review WHERE filters.
4. Confirm date range and store coverage.

# AI: Myth or Fact?

### Q1. "AI can write SQL from an English question."

**Answer:** Fact ✅

Modern AI tools can translate business questions into SQL.

Example:

Question:
"What are the top 10 stores by revenue?"

AI may generate:

```sql
SELECT store_id,
       SUM(spend_hkd) AS total
FROM sales
GROUP BY store_id
ORDER BY total DESC
LIMIT 10;
```

---

### Q2. "If AI's SQL runs, it's right."

**Answer:** Myth ❌

A query can run successfully and still produce the wrong answer.

Examples:

- Wrong JOIN
- Missing rows
- Incorrect WHERE filter
- Using total revenue instead of revenue per store

Run ≠ Correct.

---

### Q3. "AI knows that Kai Tak is missing from your store list."

**Answer:** Myth ❌

AI can only work with the data it sees.

If the reference table is missing store S09 (Kai Tak), AI does not automatically know that the data is incomplete.

AI cannot discover missing business context unless evidence exists in the data.

---

## Key Lesson

```text
AI can generate SQL.

AI cannot validate your business logic for you.

The analyst must still:
- check joins
- validate totals
- understand metrics
- question assumptions
```

# The Same Mistake, Four Ways

| Mistake | Python | R | SQL |
|----------|----------|----------|----------|
| Lost rows in a join | `how="inner"` | `inner_join()` | `INNER JOIN` |
| Empty treated as a value | `== None` | `== NA` | `= NULL` |
| Wrong quotes | works | works | `"S01"` breaks |

## Key Lesson

```text
Not all mistakes produce an error message.

Lost rows in a JOIN,
incorrect NULL handling,
and wrong business logic
can all return incorrect answers without showing red text.

Always validate totals after every JOIN.
```
